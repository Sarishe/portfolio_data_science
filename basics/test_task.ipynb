# Simple but difficult to scale, also single input at a time

```python
import math # for pi number
```

```python
text = input('Please, provide shape and dims in format: \n Square TopRight 1 1 Side 1, \n Rectangle TopRight 2 2 BottomLeft 1 1, \n Circle Center 1 1 Radius 2": \n') # to collect input data
```

```python
parts = text.strip().split()
shape = parts[0].lower()
if shape == 'square':
  side = float(parts[5])
  perimeter = 4 * side
  area = side * side
  print('Square Perimeter', round(perimeter), 'Area', round(area))
elif shape == 'circle':
  radius = float(parts[5])
  perimeter = 2 * math.pi * radius
  area = math.pi * radius * radius
  print('Circle Perimeter', round(perimeter), 'Area', round(area))

elif shape == 'rectangle':
  x1 = float(parts[2])
  y1 = float(parts[3])
  x2 = float(parts[5])
  y2 = float(parts[6])
  width = abs(x1 - x2)
  height = abs(y1 - y2)
  perimeter = 2 * (width + height)
  area = width * height
  print('Rectangle Perimeter', round(perimeter), 'Area', round(area))

else:
    print(f"Shape '{parts[0]}' not supported.")
```

# Another option, would reduce human error but different input format from what you have requested

```python
options = ["Square", "Rectangle", "Circle"]

print("Select a shape:")
for i, opt in enumerate(options, 1):
    print(i, opt)

choice = input("Enter the number: ")

try:
    index = int(choice) - 1
    if 0 <= index < len(options):
        shape = options[index]
        if shape == 'Square':
            side = float(input('Please, enter side: '))
            perimeter = 4 * side
            area = side * side
            print('Square Perimeter', round(perimeter), 'Area', round(area))

        elif shape == 'Circle':
            radius = float(input('Please, enter radius: '))
            perimeter = 2 * math.pi * radius
            area = math.pi * radius * radius
            print('Circle Perimeter', round(perimeter), 'Area', round(area))

        elif shape == 'Rectangle':
            coords = input("Enter coordinates x1 y1 x2 y2 (e.g. 2 2 1 1): ").strip().split()
            if len(coords) != 4:
                print("Invalid input. Please enter exactly 4 numbers separated by spaces.")
            else:
                x1, y1, x2, y2 = map(float, coords)
                width = abs(x1 - x2)
                height = abs(y1 - y2)
                perimeter = 2 * (width + height)
                area = width * height
                print('Rectangle Perimeter', round(perimeter), 'Area', round(area))
    else:
        print("Invalid choice.")
except ValueError:
    print("Please enter a valid number.")
```

# Classes would probably be a better case here for scaling

```python
class Square:
    def __init__(self, side):
        self.side = side
    def perimeter(self):
        return 4 * self.side
    def area(self):
        return self.side ** 2

class Circle:
    def __init__(self, radius):
        self.radius = radius
    def perimeter(self):
        return 2 * math.pi * self.radius
    def area(self):
        return math.pi * self.radius ** 2

class Rectangle:
    def __init__(self, x1, y1, x2, y2):
        self.width = abs(x1 - x2)
        self.height = abs(y1 - y2)
    def perimeter(self):
        return 2 * (self.width + self.height)
    def area(self):
        return self.width * self.height

options = ["Square", "Rectangle", "Circle"]

print("Select a shape:")
for i, opt in enumerate(options, 1):
    print(i, opt)
choice = input("Enter number: ")

if choice == '1':
    side = float(input("Enter side: "))
    shape = Square(side)
elif choice == '2':
    coords = input("Enter x1 y1 x2 y2: ").split()
    x1, y1, x2, y2 = map(float, coords)
    shape = Rectangle(x1, y1, x2, y2)
elif choice == '3':
    radius = float(input("Enter radius: "))
    shape = Circle(radius)
else:
    print("Invalid choice")

print(shape.__class__.__name__, 'Perimeter', round(shape.perimeter()), 'Area', round(shape.area()))
```
